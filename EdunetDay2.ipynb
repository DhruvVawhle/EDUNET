```python
# Odd & Even in List

lst = [1,4,6,9,8,10,11]
```

```python
even_num = []
for i in lst:
    if i % 2 == 0:
        even_num.append(i)
```

```python
even_num
```

```python
odd_num = []
for i in lst:
    if i % 2 == 1:
        odd_num.append(i)
```

```python
odd_num
```

```python
#list Comprehension

odd_num = [i for i in lst if i % 2 != 0]
```

```python
odd_num
```
