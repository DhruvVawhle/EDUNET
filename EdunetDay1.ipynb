```python
#find the square of 2 nums
#add the square of 2 nums

def square(x,y):
    return (x*x,y*y)

def sumOfsquare(x,y):
    return square(x,y)[0]+square(x,y)[1]

print(square(2,4))
print(sumOfsquare(2,4))
```
