```python
#Importing Necessary libs

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
#load the dataset

df = pd.read_csv('appliance_energy.csv')
df.head()
```

```python
#rows and columns
df.shape
```

```python
#information of the dataset
df.info()
```

```python
#statistics of the dataset

df.describe()
```

```python
#To check missing values

df.isnull().sum()
```

### Feature and Target Selection
#### X,x - Feature | y,Y - Target 

```python
X = df[['Temperature (°C)']] #it consist of multidimensional it has two square braces
y = df['Energy Consumption (kWh)'] #it consist of singledimensional it has single square braces
```

```python
# Split the dataset - 80/20 | 70/30

X_train,X_test,y_train,y_test = train_test_split(
 X,y, train_size=0.8, random_state=42
)
```

```python
X_train
```

```python
#Instantiate the Model

model = LinearRegression()
```

```python
#Train the model

model.fit(X_train, y_train)
```

```python
# let us make prediction

y_pred = model.predict(X_test)
y_pred
```

```python
# calculate the mean squared error

mse = mean_squared_error(y_test,y_pred)
print('The Mean Square Error:',mse)
```

```python
# Performance of the function , it values always will be in percentage be the number million,billions etc

r2 = r2_score(y_test,y_pred)
print('The R Squared Score:',r2)
```

```python
plt.scatter(X_test,y_test, color='blue',label='Test Data')
plt.plot(X_test,y_pred, color='Red',linewidth='4' , label='Regression line')
plt.xlabel('Temperature')
plt.ylabel('Energy Consumption')
plt.legend()
plt.title('Energy Consumption Prediction Using Simple Linear Regression')
plt.show()
```
